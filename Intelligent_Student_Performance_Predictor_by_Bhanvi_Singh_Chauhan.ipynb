```python
#Intelligent Student Performance Predictor by Bhanvi Singh Chauhan

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import mean_squared_error, accuracy_score

url = "https://raw.githubusercontent.com/amankharwal/Website-data/master/student-mat.csv"
df = pd.read_csv(url)

df['Pass_Fail'] = (df['G3'] >= 10).astype(int)

le = LabelEncoder()
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = le.fit_transform(df[col])


X = df[['studytime', 'failures', 'absences', 'Medu', 'Fedu', 'goout']]
y_reg = df['G3']
y_clf = df['Pass_Fail']


X_train, X_test, y_train_reg, y_test_reg = train_test_split(X, y_reg, test_size=0.2, random_state=42)
X_train_c, X_test_c, y_train_clf, y_test_clf = train_test_split(X, y_clf, test_size=0.2, random_state=42)


lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train_reg)
y_pred_reg = lin_reg.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))

log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_c, y_train_clf)
y_pred_log = log_reg.predict(X_test_c)
log_acc = accuracy_score(y_test_clf, y_pred_log)


dt_clf = DecisionTreeClassifier(criterion='entropy', max_depth=3)
dt_clf.fit(X_train_c, y_train_clf)
y_pred_dt = dt_clf.predict(X_test_c)
dt_acc = accuracy_score(y_test_clf, y_pred_dt)

print("\n--- PERFORMANCE ---")
print(f"Linear Regression RMSE: {rmse:.2f}")
print(f"Logistic Regression Accuracy: {log_acc*100:.2f}%")
print(f"Decision Tree Accuracy: {dt_acc*100:.2f}%")


plt.scatter(y_test_reg, y_pred_reg, color='teal')
plt.plot([y_test_reg.min(), y_test_reg.max()], [y_test_reg.min(), y_test_reg.max()], 'r--')
plt.title("Actual vs Predicted Marks")
plt.show()
```

**Objective:** "I developed an Intelligent Student Performance Predictor. My goal was to see if AI can predict a student's final grade based on factors like study time, previous failures, and social activities."
Hybrid Approach: "I implemented both Regression and Classification models.
I used Linear Regression to predict the exact marks (Continuous value).
I used Logistic Regression and Decision Trees to predict the categorical result (Pass or Fail)."

**Data Pre-processing:** "I used Label Encoding to convert categorical text data into numbers and used Feature Selection to pick the variables that most affect a student’s success."

**Key Insight:** "I found that 'Previous Failures' and 'Study Time' were the most significant features. My Decision Tree model achieved around [X]% accuracy, showing that these patterns are highly predictable."
